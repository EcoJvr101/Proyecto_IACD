```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

#Carga de datos
df = pd.read_csv('../data/processed_data.csv')
print(f'Dataset listo para analizar. Filas: {df.shape[0]}')
```

```python
#Mapa de calor, que factores estan relacionados con la perdida de peso?
plt.figure(figsize=(10, 6))
numericas = df.select_dtypes(include=['float64', 'int64'])
correlacion = numericas.corr()

sns.heatmap(correlacion, annot=False, cmap='coolwarm', center=0)
plt.title('Mapa de Correlación: Factores de la Dieta y Paciente')
plt.show()
```

**Conclusión de la Correlación:**
Al observar el mapa de calor, notamos que la variable `weight_change_kg_6m` tiene una fuerte relación con X y con Y. Esto nos indica que el perfil inicial del paciente importa más que la restricción de calorías pura.

```python
# Que dieta genera mayor perdida de pero
plt.figure(figsize=(8,6))
sns.boxplot(data=df, x='diet_type', y = 'weight_change_kg_6m')
plt.title('Distribución de Pérdida de Peso por Tipo de Dieta')
plt.ylabel('Cambio de Peso a los 6 meses (kg)')
plt.show()
```

**Análisis de Dietas:**
El gráfico de cajas (boxplot) revela las diferencias en la distribución de la pérdida de peso según el tipo de dieta. Aquellas dietas con una mediana más baja (valores más negativos) indican una mayor eficacia general en la reducción de peso. Además, la amplitud de las cajas y los "bigotes" nos muestran la variabilidad de respuesta de los pacientes a cada tipo de restricción.
Celda 6: Código (Motivación vs Resultados)

```python
#Afecta la motivacion al exito a largo plazo?
plt.figure(figsize=(8,6))
# scatterplot coloreando por sexo para buscar sub-patrones
sns.scatterplot(data=df, x='motivation_score', y='weight_change_kg_6m', alpha=0.6, hue='sex')
plt.title('Impacto de la Motivación Inicial en la Pérdida de Peso')
plt.ylabel('Cambio de Peso a los 6 meses (kg)')
plt.xlabel('Puntuación de Motivación Inicial (0 a 1)')
plt.axhline(0, color='red', linestyle='--', linewidth=1) # Línea de 0kg perdidos
plt.show()
```

**Análisis Conductual:**
El gráfico de dispersión relaciona el estado psicológico inicial del paciente (`motivation_score`) con su pérdida de peso real. Observamos si existe una tendencia de que los valores más altos de motivación se agrupen hacia la parte inferior del eje Y (mayor pérdida de peso).
